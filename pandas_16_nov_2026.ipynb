```python
pd_series2 = pd.Series(data=[1776, 1867, 1821], index=['USA', 'CANADA', 'MEXICO'])
print(pd_series2.values, end='\n\n')
print(pd_series2.keys, end='\n\n')
print(pd_series2, end='\n\n')
print(pd_series2['USA'])
print(pd_series2['MEXICO'])
pd_series2['ISRAEL'] = 1948
print(pd_series2['ISRAEL'], end='\n\n')

print(pd_series2.iloc[3])  # index - location only!

print(pd_series2.loc['CANADA':'MEXICO'], end='\n\n')
print(pd_series2['CANADA':'MEXICO'])
```

```python
first_series = pd.Series({'Japan': 100, 'China': 450, 'USA': 250})
second_series = pd.Series({'Brazil': 70, 'China': 350, 'USA': 200})

concat_series = first_series + second_series

print(concat_series)
```

```python
first_series = pd.Series({'Japan': 100, 'China': 450, 'USA': 250})
second_series = pd.Series({'Brazil': 70, 'China': 350, 'USA': 200})

concat_series = first_series.add(second_series, fill_value=0)

print(concat_series)
```

```python
second_series = pd.Series({'Brazil': 70.3, 'China': 350.9, 'USA': 200.0})
print(second_series, end='\n\n')

second_series.astype(int)
```

```python
first_series = pd.Series({'Japan': 100, 'China': 450, 'USA': 250})
second_series = pd.Series({'Brazil': 70, 'China': 350, 'USA': 200})

print(first_series[first_series > 400], end= '\n\n')

print(first_series[[False, True, False]], end='\n\n')

print(first_series[pd.Series({'Japan': True, 'China': False, 'USA': True})], end='\n\n')

print(first_series[pd.Series({'Japan': True, 'USA': True, 'China': False })], end='\n\n')
```
